```python
i = 0

while (i<=10):
    print(i)
    i += 1    # i = i+1
```

### Nested Loop

```python
from IPython.display import Image
Image('Nested Loop Build this Pyramid.png')
```

```python
rows = 4
# outer loop
for x in range(1, rows+1):
    # inner loop
    for y in range(1, x+1):
        print('#', end=' ')
    print('')
```

```python
num = [1, 2, 3, 4]

# outer loop
for ele in num:
    # inner loop
    index = 0
    while index < len(num):
        print(num[index], end = ' ')
        index += 1
    print('\n')
```
