### **Calculate Factorial Number using Function**

```python
# Using for loop

def fact(x):

  result = 1

  if x == 1 or x == 0:
    print(f'The {x}! is = 1')

  else:
    for i in range(x):
      result = result * (i+1)

    return result

fact(5)
```

```python
# Using while loop

def fact(x):

    result = 1

    if x == 0 or x == 1:
        print(f'The {x}! is = 1')
        return 1  # return value for consistency

    else:
        i = 1
        while i <= x:
            result *= i    # result = result * i
            i += 1         # i = i + 1
        return result

print(fact(5))
```
