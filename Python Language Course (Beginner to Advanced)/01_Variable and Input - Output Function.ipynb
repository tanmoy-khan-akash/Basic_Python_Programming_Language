# Global Variable
#### Outside of a function

```python
x = 2500        # Global Variable     # Global Variable has more power here.

def fun(): 
    print('My number is: ', x)
fun()
```

# Local Variable
#### Inside of a function

```python
x = 2500        # Global Variable

def fun():    
    x = 1000    # Local Variable     # Local Variable has more power here.
    print('My number is: ', x)
fun()
```

```python
x = 2500        # Global Variable

def fun():    
    x = 1000    # Local Variable     # Local Variable has more power here.
    print('My number is: ', x)   # Here, x is  inside the function
fun()

print('My number is: ', x)   # Here, x is outside the function
```

```python
# a = 500        # Global Variable

def fun():    
    a = 200    # Local Variable     # Local Variable has more power here.
    print('My number is: ', a)   # Here, a is  inside the function
fun()

print('My number is: ', a)   # Here, a is outside the function
```

```python
def fun(): 
    global u    # Here, the local variable u is declared as a global variable 
    u = 1000    # Here, u is a local variable 
    print('My number is: ', u)   # Here,  u is  inside the function
fun()

print('My number is: ', u)   # Here, u is outside the function
```
