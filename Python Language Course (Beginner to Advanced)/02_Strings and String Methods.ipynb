# Python String Methods

##### Find 

```python
data = 'Python Language'
data.find('La')
```

```python
data.find('o')   # Find value from string
```

```python
data.find('u',6,10)   # Not exist   # If the value is not within the range in Find, the arrow will show -1.
```

```python
data.find('g',6,14)
```

##### LOWER CASE

```python
data.lower()   # Convert all char into LOWER case
```

```python
data.casefold()   # Convert all char into LOWER case
```

##### UPPER CASE

```python
data.upper()   # Convert all char into UPPER case
```

##### Only first char into UPPER case

```python
data.capitalize()   # Convert only first char into UPPER case
```

```python
data = 'pYthon LanGuagE'   # Convert only the first character into UPPER case
data.capitalize()
```

##### Only first char into LOWER case

```python
data = 'Python Language'
data.casefold()   # Convert only first char into LOWER case
```

##### Swape case

```python
data = 'python LANGUAGE'   # UPPER case doing LOWER case convert & # LOWER case doing UPPER case
data.swapcase()
```

##### Title case

```python
data = 'python LANGUAGE'   # Only first char UPPER case
data.title()
```

##### IS LOWER case

```python
data = 'python LANGUAGE'   
data.islower()
```

##### IS UPPER case

```python
data = 'python LANGUAGE'   
data.isupper()
```

##### ENCODE case

```python
data = 'python LANGUAGE'   # Show output Binary value
data.encode()
```

```python
type(data.encode())
```
