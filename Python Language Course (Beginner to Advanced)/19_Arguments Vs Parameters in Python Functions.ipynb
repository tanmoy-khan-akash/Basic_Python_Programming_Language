### **Parameters Vs. Arguments in Python Functions**

#### 1. Parameters: These are the names listed in the function definition. They act as placeholders for the actual values that will be passed into the function when it is called.
#### 2. Arguments: These are the actual values that are passed to a function when it is called. Arguments are the values that replace the parameters when the function is invoked.

```python
def add_num(x, y):      # Parameters (x, y)
  result = x + y

  return result

add_num(10, 20)         # Arguments (10, 20)
```

```python
def add_num(x, y):
  result = x * y

  return result

add_num(10, 20)
```

```python
def add_num(x, y):
  print(x)

add_num(10, 20)
```

### **1. Positional Argument**
### **2. Keyword Argument**
### **3. Default Argument**
### **4. Variable Length Argument**

#### **Positional Argument**

```python
def add_num(x, y):
  result = x + y

  return result

add_num(10, 20)
```

#### **Keyword Argument**

```python
def personal(job, income):
  print(job)
  print(income)

personal('Data Science', 20000)
```

```python
def personal(job, income):
  print('job = ', job)
  print('salary = ', income)

personal(income = 20000, job = 'Data Scientist')
```

#### **Default Argument**

```python
def personal(job, income = 10000):
  print('job = ', job)
  print('salary = ', income)

personal('Data Scientist', 20000)          

# 10000 default argument    # value assign = 20000 
# At the output, the default will not come, Value assign will come = 20000
```

#### **Variable Length Argument**

```python
def add_num(*x):
  print(x)
  result = sum(x)
  return result

add_num(10, 20, 30, 40, 50)
```

```python
def add_num(x, *y):
  print(x)
  print(y)

add_num(10, 20, 30, 40, 50)
```

#### **Arguments** and **Keyword_Arguments**

```python
def personal_data(name, *arg):
  print(name)
  print(arg)

personal_data('Tanmoy Khan Akash', 'WUB', 'Dhaka')    # Tuple will show the output alone
```

```python
def personal_data(name, **arg):
  print(name)
  print(arg)

personal_data('Tanmoy Khan Akash', versity = 'WUB', city = 'Dhaka')   # Dictionary will show the output alone
```
