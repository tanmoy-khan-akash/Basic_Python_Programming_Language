### Frozen Set in Python

```python
from IPython.display import Image
Image('Frozen Set.png')
```

```python
s1 = {1,2,3,4,5}
s2 = {'a','b'}
```

```python
s1
```

```python
s2
```

```python
type(s1)
```

```python
s1.add(10)
s1
```

```python
# FrozenSet

s3 = frozenset(s1)
s3
```

```python
type(s3)
```

```python
# frozenset can not: add / modify / remove

s3.add(11)
s3
```

```python
# frozenset can not: add / modify / remove

s3.remove(10)
s3
```

```python
li = [1,2,3,4,15]
s4 = frozenset(li)
s4
```

```python
s4[0]
```

```python
li[0]
```

```python
s3
```

```python
s4
```

```python
# union

s3.union(s4)
```

```python
# intersection

s3.intersection(s4)
```

```python
s = 'data science'
s
```

```python
s5 = frozenset(s)
s5
```

```python
[tuple(x) for x in s5]
```

```python
[set(x) for x in s5]
```
