### List Comprehension

```python
from IPython.display import Image
Image('List Comprehension.png')
```

```python
from IPython.display import Image
Image('Simple List Comprehension.png')
```

```python
squares = [x**2 for x in range(10)]
print(squares)
```

```python
[x for x in range(10)]
```

```python
for x in range(10):
    print(x)
```

```python
from IPython.display import Image
Image('List Comprehension with Condition.png')
```

```python
evens = [x for x in range(10) if x%2 == 0]
print(evens)
```

```python
data = 'I love Data Science'

[x for x in data]
```

```python
data = 'I love Data Science'

[x for x in data if x != 'I']
```

```python
data = 'I love Data Science'
data = data.split()
[x for x in data if x != 'I']
```

```python
matrix = [[j for j in range(5)] for i in range(3)]
print(matrix)
```
