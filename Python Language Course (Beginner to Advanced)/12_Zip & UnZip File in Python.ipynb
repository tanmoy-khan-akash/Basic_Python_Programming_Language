#### Duplicate ID will not be accepted.

```python
# Zip function

name = ['Tanmoy', 'Usha', 'Sishir']
versity = ['WUB', 'DMC', 'SUST']
```

```python
# List

zipped = list(zip(name, versity))
zipped
```

```python
# Tuple

zipped = tuple(zip(name, versity))
zipped
```

```python
# Dictionary

zipped = dict(zip(name, versity))
zipped
```

```python
zipped.keys()
```

```python
zipped.values()
```

```python
# Set   # Duplicate keys aren't allowed 

zipped = set(zip(name, versity))
zipped
```

### UnZip

```python
name = ['Tanmoy',' Usha', 'Sishir']
versity = ['WUB', 'DMC', 'SUST']
```

```python
zipped = list(zip(name, versity))
zipped
```

```python
# UnZip

name, versity = zip(*zipped)
```

### Zip & UnZip File with 2 Lines of Python Code

```python
# Zip File Python Code:

import zipfile as zp

zipp = zp.ZipFile('Zipped Rename.zip', 'w')     # 'w' means stands for write

zipp.write('Tanmoy_Khan_Akash.jpg')
zipp.write('Water_Pot.png')
zipp.write('Water Pot.jpeg')
zipp.write('Example.rtf')
```

```python
# UnZip File Python Code:

with zp.ZipFile('Zipped Rename.zip', 'r') as file:
    file.extractall('UnZip Zipped Rename')
```

### Unpacking in Python

```python
ids = (1,2,3)
name = ('Tanmoy','Usha','Sishir')
```

```python
zipped = tuple(zip(ids, name))
zipped
```

```python
ids_new, name_new = zip(*zipped)
```

```python
ids_new
```

```python
name_new
```

```python
i1, n1 = zipped[0]
i1
```

```python
n1
```

```python
zipped
```

```python
# Unpacking using loop

for ids, name in zipped:
  print(ids, name)
```

```python
# Or

#Enumerating: Iterate over pairs 
# Unpacking iterate over parts

for data in enumerate(zipped):
  # Unpack pair: ids, name
  ids, name = data
  print(data)
```

```python
for index, pair in enumerate(zipped):
  # Unpack pair: ids, name
  ids, name = pair
  print('Rank {}, id = {}, name = {}'. format(index, ids, name))
```
