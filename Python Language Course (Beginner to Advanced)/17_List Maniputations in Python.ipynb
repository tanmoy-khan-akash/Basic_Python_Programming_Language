### **Question:**

#### Create a function named remove_duplicates. It will take a list (input_list) as its parameter. It then iterates through the elements of the input list and appends each unique element to a new list (unique_list). This new list is then returned.

```python
def remove_duplicates(input_list):
  unique_list = []

  for item in input_list:
    if item not in unique_list:
      unique_list.append(item)

  return unique_list

input_list = [1,2,3,4,2,3,2,7,8]
remove_duplicates(input_list)
```

```python
print(f'Real list {input_list}')
print(f'Unique list {remove_duplicates(input_list)}')
```

```python
def double_elements(input_list):
  doule_list = [2*x for x in input_list]

  return doule_list

input_list = [1,2,3,4,2,3,2,7,8]
double_elements(input_list)
```
