```python
markes = float(input('Enter your markes: '))

if markes < 25:
    print('F Grade, You are fail')
elif markes >= 25 and markes <= 44:
    print('E Grade, Pass')
elif markes >= 45 and markes <= 49:
    print ('D Grade, Pass')
elif markes >= 50 and markes <= 59:
    print('C Grade, Pass')
elif markes >= 60 and markes <= 69:
    print('B Grade, Pass')
elif markes >= 70 and markes <= 79:
    print('A- Grade, Pass')
elif markes >= 80 and markes <= 89:
    print('A Grade, Pass')
    
else:
    # markes >= 90 and markes = 100
    print('A+ Grade, Pass')
```
