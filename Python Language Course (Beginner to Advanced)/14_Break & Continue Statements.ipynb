### Break & Continue Statements in Python

#### Break Statements:

```python
from IPython.display import Image
Image('Break.png')
```

```python
# Break
# For the For loop

for i in range(20):
  if i == 10:
    break
  print(i)
```

```python
for i in range(1, 20):
  if i == 11:
    break
  print(i)
```

```python
for i in range(1, 20):   # Even
  #if i == 11:
  if i % 2 == 0:
    #break
    print(i)
```

```python
for i in range(20):
  if i == 11:
    break
  print('3 *', i, '=', i*3)
```

```python
for i in range(20):
  if i == 11:
    break
  print('5 *', i, '=', i*5)
```

```python
# In the case of Strings

sent = 'I love Bangladesh and USA'
```

```python
for i in sent.split():
  if i == 'and':
    break
  print(i)
```

```python
# Break
# For the While loop

i = 1

while i <= 100:
  if i == 11:
    break
  print('5 *', i, '=', 5*i)
  i += 1
```

#### Continue Statements:

```python
from IPython.display import Image
Image('Continue.png')
```

```python
num = [1,2,3,4,5,10,15,20,25,30,35,40,45,50]
num
```

```python
i = 1

for i in num:
  if i%2 == 0:
    print(i)
```

```python
# List comparation

[i for i in num if i%2 == 0]    # Even
```

```python
num = [1,2,3,4,5,10,15,20,25,30,35,40,45,50]
num
```

```python
# Continue
# For the For loop

for i in num:
  if i%2 == 1:    # Even
    continue
  print(i)
```

```python
for i in num:
  if i % 2 == 0:    # Odd
    continue
  print(i)
```

```python
# In the case of String

sent = 'I love Bangladesh and USA'
```

```python
for i in sent.split():
  if i == 'and':
    continue
  print(i)
```

```python
i = 1
while i <= 10:
  if i == 5:
    continue
  print('5 *', i,'=', 5*i)
  i += 1
```

```python
i = 1
while i <= 10:
  if i * 2 == 11:
    continue
  print('5 *', i,'=', 5*i)
  i += 1
```
