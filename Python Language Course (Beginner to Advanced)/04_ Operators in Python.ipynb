# Operators in Python 

##### 1. Arithmetic Operators:     
* (+, -, *, /, %, **, //)
##### 2. Identity Operators:     
* (is, is not)
##### 3. Bitwise Operators:     
* (& (AND), | (OR), ~ (NOT), << (Left Shift), >> (Right Shift))
##### 4. Comparison Operators:     
* (==, !=, >, <, >=, <=)
##### 5. Logical Operators:     
* (and, or, not)
##### 6. Assignment Operators:     
* (=, +=, -=, *=, /=, //=, %=, **=, &=, |=, ^=, >>=, <<=)
##### 7. Membership Operators:     
* (in, not in)
