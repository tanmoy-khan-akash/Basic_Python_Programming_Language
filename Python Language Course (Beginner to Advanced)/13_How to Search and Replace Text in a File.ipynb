### How to Search and Replace Text in a File in Python

```python
search_word = 'Bangladesh'
replace_word = 'Barcelona'

with open(r'Example.rtf', 'r') as file_ref:
  data = file_ref.read()
  data = data.replace(search_word, replace_word)
```

```python
with open(r'Example.rtf', 'w') as file_ref:
  file_ref.write(data)
```

### If I want to do it directly instead of from a file

```python
text = 'Our country name is Barcelona. It is a captivating country.'
```

```python
search_word = 'captivating'
replace_word = 'charming'
```

```python
import re    # re means regular expression

re.sub(search_word, replace_word, text)
```
